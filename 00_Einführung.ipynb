Ziel: Beziehung zwischen einer Zielvariablen $Y$ und mehreren Eingabevariablen $X = (X_1, X_2, \dots, X_p)$ als mathematische Funktion $f$ darzustellen:

$$ Y = f(X) + \epsilon $$

mit dem zufälligen, unkorrelierten Fehlerterm $\epsilon$, der durch das Modell nicht erklärt werden kann.

Zwei große Ziele:

* Muster erkennen (Erklärung)
* Vorhersagen treffen (Prognose)

Typische Methoden

* lineare Regression   $ Y = \beta_{0} + \beta_{1} X_{1} + \epsilon $
* multiple Regression  $ Y = \beta_{0} + \beta_{1} X_{1} + \beta_{2} X_{2} + ... + \beta_{p} X_{p} + \epsilon $
